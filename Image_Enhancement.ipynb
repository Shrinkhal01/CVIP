```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read image in grayscale
img = cv2.imread("image2.jpg", cv2.IMREAD_GRAYSCALE)

# 1. Negative transformation
negative = 255 - img

# 2. Brightness adjustment
bright = cv2.convertScaleAbs(img, alpha=1.0, beta=50)

# 3. Contrast enhancement
contrast = cv2.convertScaleAbs(img, alpha=1.5, beta=0)

# 4. Histogram Equalization
equalized = cv2.equalizeHist(img)
```

```python
# Display results
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(negative, cmap="gray")
plt.title("Negative")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(bright, cmap="gray")
plt.title("Brightness Enhanced")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(contrast, cmap="gray")
plt.title("Contrast Enhanced")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(equalized, cmap="gray")
plt.title("Histogram Equalized")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Display histograms
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(img.ravel(), 256, [0, 256])
plt.title("Original Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(equalized.ravel(), 256, [0, 256])
plt.title("Equalized Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# Save enhanced image
cv2.imwrite("enhanced_image.jpg", equalized)

print("Enhanced image saved successfully.")
```

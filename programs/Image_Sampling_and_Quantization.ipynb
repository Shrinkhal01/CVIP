```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Read image in grayscale
img = cv2.imread("image1.jpg", cv2.IMREAD_GRAYSCALE)

# Reduce image resolution
small = cv2.resize(img, (128, 128))

# Resize using nearest neighbour
nearest = cv2.resize(small, (512, 512),
                     interpolation=cv2.INTER_NEAREST)

# Resize using bilinear interpolation
bilinear = cv2.resize(small, (512, 512),
                      interpolation=cv2.INTER_LINEAR)
```

```python
# Function for gray-level quantization
def quantize(image, levels):
    return (image // (256 // levels)) * (256 // levels)

levels = [256, 128, 64, 32, 16, 8]
quantized = [quantize(img, level) for level in levels]
```

```python
# Display sampling results
plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(nearest, cmap="gray")
plt.title("Nearest Neighbour")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(bilinear, cmap="gray")
plt.title("Bilinear")
plt.axis("off")

plt.show()
```

```python
# Display quantization results
plt.figure(figsize=(12, 6))

for i in range(len(levels)):
    plt.subplot(2, 3, i + 1)
    plt.imshow(quantized[i], cmap="gray", vmin=0, vmax=255)
    plt.title(f"{levels[i]} Gray Levels")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
